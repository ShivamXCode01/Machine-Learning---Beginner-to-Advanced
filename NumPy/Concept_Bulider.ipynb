**ARRAY OPERATIONS**

***Array Slicing***

*We can get the certain part of an array using slicing function*
*Some examples are give for 1D , 2D ,ND arrays*

```python
One_D =np.arange(9)
One_D
```

```python
print("Basic Slicing of array:",One_D[3:7])
```

```python
print("Basic Slicing of array:",One_D[2:])
```

```python
print("Basic Slicing of array:",One_D[:4])
```

*In the above examples we can easily see that we can provide starting index and ending index(Exclusive) during the slicing if we not provide the starting index then it start from 0th index and if you not provide ending index then it goes till the length of the array*

```python
print("Slicing with step:",One_D[0:6:2])
```

*In the above example we slicing the array using step method in we passes three argument first for starting index and second for ending index and for step means how much we jump one element to another element* -----------We can also do negative index the example is given below 

```python
print("Negative Indexing:",One_D[-1])
```

*In the below examples we are learning different types of slicing 2D array*

```python
Two_D = One_D.reshape((3,3))
Two_D
```

```python
print("slicing for specific element:",Two_D[2,1])
```

```python
print("Slicing for entire row:",Two_D[0])
```

```python
print("Slicing for entire column:",Two_D[:,2])
```

```python
print("slicing for specific portion of matrix:",Two_D[1:3,1:3])
```

***Sorting in arrays***

```python
unSorted =np.array([4,21,8,0,3,1,2,7])
unSorted
```

```python
sorted = np.sort(unSorted)
sorted
```

```python
unSorted_2D =unSorted.reshape((4,2))
unSorted_2D
```

*By default in 2D sorting perfform row was if we need to change this we have to give the axis= 1 or 0*-------------------*1 for row and 0 for column we can see the the given example*

```python
sorted_2D = np.sort(unSorted_2D)
sorted_2D
```

```python
sorted_2D = np.sort(unSorted_2D,axis=0)
sorted_2D
```

**We can also perform some Linear Algebric functions with array with the help of numpy**

```python
al = np.arange(16)
print(al)
al1 = np.arange(1,33,2)
print(al1)
```

```python
al2 =al.reshape((4,4))
print(al2)
al3 = al1.reshape((4,4))
print(al3)
```

```python
print(np.linalg.det((al3,al2)))
```

```python
print(np.linalg.matrix_rank((al2,al3)))
```

```python
print(np.linalg.solve(al2,al3))
```

```python
print(np.linalg.inv(al3))
```

```python
print(np.linalg.matrix_transpose(al2))
```

```python
print(np.linalg.diagonal(al2))
```
