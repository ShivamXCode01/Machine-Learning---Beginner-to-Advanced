**Univariate means analyzing single variable (single column analysis independently)**

*There are two types of data 1.Categorical (Country,colour etc)  2.Numerical (Age,Height,Weight etc)**

```python
import numpy as np
import pandas as pd
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

**1.Categorical Data Handling**

```python
df = pd.read_csv("Titanic-Dataset.csv")
df
```

```python
df.info()
```

```python
#Count plot :- It gives the frequency of each categories present in the column
sns.countplot(df) # it gives all the feature
sns.countplot(df["Sex"],color="blue")
```

```python
df["Sex"].value_counts().plot(kind='bar') # gives the values counts of each features according to their category
```

```python
#PieCharts:- It gives the information in the form of piechart
df['Survived'].value_counts().plot(kind='pie',autopct='%.2F')
```

**handling numerical Features**

```python
plt.hist(df['Age'],bins=15)
```

```python
df['Age'].plot(kind='hist',bins=15)
```
