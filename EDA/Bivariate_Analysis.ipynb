**1.Numerical-Numerical**:- If we have two columns of numerical dataTypes 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
df
```

```python
df.info()
```

```python
# it is used when we have numerical-numerical data
sns.scatterplot(x='Age',y='Fare',data=df,hue='Sex')
```

```python
# it is used when we have categorical- numerical data
sns.barplot(x='Sex',y='Age',data=df,color='orange')
```

```python
sns.boxplot(x='Sex',y='Age',data=df,color='grey')
```

```python
# it is used if we have categorical-categorical data 
sns.heatmap((df['SibSp'],df['Survived']))
```

```python
sns.pairplot(df,hue='Sex') # it gives the all scatter with all the columns(Numerical)
```
