```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly as pt
```

```python
x = np.linspace(0,50,10)
x
```

```python
y = x ** 2
y
```

*Making plots*

```python
plt.plot(x,y)
plt.title("Graph") #it gives title pf the graph
plt.xlabel("X-axis") # it gives X-axis naming
plt.ylabel("Y-axis") # it give Y-axis naming
```

*Making Sub plots*

```python
plt.subplot(2,4,5)  # it is used to making subplots in a given canvas first parameter is for rows second for column third for selexting
                    # which region or graph we select
plt.plot(x,y)
plt.xlabel("X")
plt.ylabel("Y")
plt.subplot(2,4,3)
plt.plot(y,x)
plt.xlabel("y")
plt.ylabel("x")
```

**Object Oriented Way :-**
It is used to access the canvas functionality

```python
fig = plt.figure(figsize=(8,6))
axis1 = fig.add_axes([1,0.8,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.6,1,0.5,0.5])
axis2.plot(y,x)
```

**Types Of Plot**

```python
plt.scatter(x,y)  # it gives Scatter plot
```

```python
plt.hist(x,y)
```

```python
plt.hist2d(x,y)
plt.xlabel("X")
```

```python
plt.boxplot(x) # it gives signle argument
```

```python
plt.bar(x,y)
```

```python
plt.stackplot(x,y)
```

**Saving Plot**

```python
fig = plt.figure()
ax = fig.add_axes([0.8,0.8,1,1])
ax.plot(x,y)
plt.savefig("BasicPlot.png")
```

**SeaBorn Starts :- ------------------------------ Distribution Plot---------------------------------------------**

```python
df = sns.load_dataset('tips')
df
```

```python
df.shape
```

```python
df['size'].unique()
```

```python
df.sample(5)
```

***i) HistPlot / DistPlot :-*** this plot is used for Quantative Variable and it take only one value at a time

```python
sns.displot(df['total_bill'],bins=40)
```

**Joint Plot:-** it is used to plot/graph using multiple values at the same time

```python
sns.jointplot(x='total_bill',y='tip' ,data=df ,hue='day',kind='hist')
```

```python
sns.jointplot(x='total_bill',y='tip' ,data=df ,kind='reg')
```

```python
sns.jointplot(x='total_bill',y='tip' ,data=df,kind='resid')
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind='hex',color='orange')
```

**PairPlot:-** it is used to do the comparision b/w all the numerical/quantatity data in the dataset.

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue='time')
```

```python
sns.pairplot(df,palette='rainbow',hue='day')
```

**Rug Plot:-** it gives all the numerical value in a graph

```python
sns.rugplot(df)
```

**---------------------------------------------------CATEGORICAL PLOT------------------------------------------------------------** it is used to operations on the categorical (qualatative) values present in the dataset

**i)Count Plot:-** it gives the count of single catrgorical data

```python
sns.countplot(x=df['sex'],hue=df['smoker'])
```

**ii)Bar Plot:-** it gives the statical values b/w categorical data and numerical data

```python
sns.barplot(x=df['sex'],y=df['tip'],hue=df['sex']) # by default it gives mean value
```

**iii)BoxPlot:-** it gives five number data of dataset

```python
sns.boxplot(x='tip',hue='sex',data=df,palette='rainbow')
```

**iv)ViolinPlot:-** 

```python
sns.violinplot(x='tip',y='sex',data=df)
```

**v)StripPlot && SwarmPlot:-**

```python
sns.stripplot(x='tip',y='sex',data=df)
```

```python
sns.swarmplot(x='tip',y='sex',data=df)
```

**----------------------------------------------MatrixPlot----------------------------------**

**i)HeatMap:-** it gives correralation between dataS

```python
tip_corr =df[['total_bill','size','tip']]
```

```python
sns.heatmap(tip_corr.corr(),annot=True)
```

**ClusterMap:-** 

```python
!pip show scipy
!pip install scipy
import scipy as sc
```

**Pivot table Heat Map**

```python
pvttips = df.pivot_table(values='total_bill',index='sex',columns='day')
pvttips
```

```python
sns.heatmap(pvttips)
```

**Regression Plot:-**  it is used in machine learning

```python
sns.lmplot(x='total_bill',y='tip',data=df,hue='smoker',markers=['o','v'])
```
