**One hot Encoding:-** it is used to handle the categorical data but in this their is no realtionship or order or pattern in the data

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('cars.csv')
df.head()
```

```python
df['brand'].value_counts()
```

```python
df['brand'].nunique()
```

1. OneHotEncoding using Pandas

```python
pd.get_dummies(df,columns=['fuel','owner'],dtype=int)
```

2. K-1 OneHotEncoding

```python
pd.get_dummies(df,columns=['fuel','owner'],drop_first=True,dtype=int)
```

3. OneHotEncoding using Sklearn

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0:4],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
```

```python
X_train_new=ohe.fit_transform(X_train[['fuel','owner']]).toarray()
```

```python
X_test_new = ohe.transform(X_test[['fuel','owner']]).toarray()
```
