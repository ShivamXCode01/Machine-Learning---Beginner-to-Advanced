## Discretization(Binning) 
It is the process of converting continuous variable into discrete variable by creating a set of contiguous interval that span the range of the variable's values.

Why?
1. To handle Outliers
2. To improve the value spread

## Binning


```python
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.compose import ColumnTransformer
```

```python
kbin_age = KBinsDiscretizer(n_bins=15,encode='ordinal',strategy='quantile')
kbin_fare = KBinsDiscretizer(n_bins=15,encode='ordinal',strategy='quantile')
```

```python
trf = ColumnTransformer([
    ('first',kbin_age,[0]),
    ('second',kbin_fare,[1])
])
```

```python
X_train_trf = trf.fit_transform(X_train)
X_test_trf = trf.fit_transform(X_test)
```

```python
trf.named_transformers_['first'].bin_edges_
```

```python
output = pd.DataFrame({
    'age':X_train['Age'],
    'age_trf':X_train_trf[:,0],
    'fare':X_train['Fare'],
    'fare_trf':X_train_trf[:,1]
})
```

```python
output['age_labels'] = pd.cut(x=X_train['Age'],
                                    bins=trf.named_transformers_['first'].bin_edges_[0].tolist())
output['fare_labels'] = pd.cut(x=X_train['Fare'],
                                    bins=trf.named_transformers_['second'].bin_edges_[0].tolist())
```
