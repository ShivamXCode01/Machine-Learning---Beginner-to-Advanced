### Arbitary Imputation

In this technique we fill the entire missing value with the single value that inform the model that their is missing value.
Ex:- numerical missing value(-1,0,99.99 value that not presaent in the data) , categorical missing value(missing)

When to use ?
1. Not data missing at random (missing data in chuncks)

```python
import numpy as  np 
import pandas as pd
```

```python
df = pd.read_csv('titanic_toy.csv')
df.head()
```

```python
df.isna().mean()*100
```

```python
# filling the missing value using pandas
df['Age_filled'] = df['Age'].fillna(0) # filled with zero
```

```python
df['Fare_filled'] = df['Fare'].fillna(-1)
```
