### MEAN/MEDIAN Imputation

*In this teachnique we fill the missing value for  numerical features with the help of their mean and median values  of the features(columns). Here, we uses only single feature means only that feature where we have missing value to impute the missing value*

When to Use?
1. Mean :- if we have normal distribution data
2. Median :- if we have skewed data
3. MCAR
5. less than 5% missig data

### Using Pandas

```python
Fare_mean = df['Fare'].mean()
Fare_median = df['Fare'].median()
Fare_mean
```

```python
df['Fare_filled_mean'] = df['Fare'].fillna(Fare_mean)
df['Fare_filled_median'] = df['Fare'].fillna(Fare_median)
df.sample()
```

### Using Scikit Learn

```python
from sklearn.impute import SimpleImputer
si1 = SimpleImputer(strategy='mean')
si2 = SimpleImputer(strategy='median')
```

```python
from sklearn.compose import ColumnTransformer 
```

```python
trf = ColumnTransformer([
    ('si1',si1,['Age']),
    ('si2',si2,['Fare'])]
    ,remainder='passthrough')
```

```python
df['Fare_trf'] = si1.fit_transform(df[['Fare']])
df['Fare_trf'] = si2.fit_transform(df[['Fare']])
```
