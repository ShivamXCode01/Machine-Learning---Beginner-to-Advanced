### Handling categorical missing data

In this technique there are two ways to fill the missing value 
1. Fill with the frequent value
2. Fill with new category like (missing)

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('train_2.csv')
df = pd.read_csv('train_2.csv',usecols=['GarageQual','FireplaceQu','SalePrice'])
df.head()
```

```python
df.isna().mean()*100
```

```python
# finding the columns which has 5% missing data
cols = [var for var in df.columns 
        if df[var].isna().mean() < 0.06 and df[var].isna().mean() > 0.0]
cols
```

```python
df.isna().sum()
```

```python
df['GarageQual'].value_counts().sort_values(ascending=False)
```

```python
df['FireplaceQu'].value_counts().sort_values(ascending=False)
```

```python
# filling the missing value wit TA because it is more frequent
df['GarageQual_filled'] = df['GarageQual'].fillna('TA')
```

#### Using scikit learn

```python
from sklearn.impute import SimpleImputer
si = SimpleImputer(strategy='constant',fill_value='Missing')
```

```python
cols = df.columns
df =  si.fit_transform(df)
df = pd.DataFrame(df,columns=cols)
df
```
