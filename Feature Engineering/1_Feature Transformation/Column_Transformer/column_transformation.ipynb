```python
# first we fill the missing values 
from sklearn.impute  import SimpleImputer
```

```python
# we can impute like that first fit then transform never do on test data due 
# to this it causes data leakage
# si = SimpleImputer()
# si.fit(X_train[['fever']])
# si.fit(X_test[['fever']])
# X_train_fever = si.transform(X_train[['fever']])
# X_test_fever =si.transform(X_test[['fever']])
# X_train_fever
```

```python
# here we give fit_tranform at same time
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[['fever']])
X_test_fever = si.fit_transform(X_test[['fever']])
X_train_fever
```

```python
X_train.sample(5)
```

```python
# we apply ordinal encodin on cough beacuse here is the realationship 
# b/w the data
oe = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough=oe.fit_transform(X_train[['cough']])
X_test_cough=oe.fit_transform(X_test[['cough']])
```

```python
oe.categories_
```

```python
# we apply one hot encoding in the gender and city 
ohe = OneHotEncoder()
# we apply ohe n-1
ohe = OneHotEncoder(
    drop='first',
    handle_unknown='ignore',
    sparse_output=False
)
```

```python
ohe.fit(X_train[['gender','city']])
X_train_city_gender = ohe.transform(X_train[['gender','city']])
X_test_city_gender = ohe.transform(X_test[['gender','city']])
X_train_city_gender.shape
```

```python
# we are extracting age from the dataFrame
X_train_age = X_train.drop(columns=['gender','fever','cough','city']).values

X_test_age = X_test.drop(columns=['gender','fever','cough','city']).values
```

```python
# now we merge all the features aftear ecodng in one data frame
X_train_transformed = np.concatenate((X_train_age,X_train_fever,X_train_city_gender,X_train_cough),axis=1)

X_test_transformed =np.concatenate((X_test_age,X_test_fever,X_test_city_gender,X_test_fever),axis=1)
```

```python
# we convert numpy array  datafame usin pandas
X_train_transformed=pd.DataFrame(X_train_transformed)
X_train_transformed.sample(7)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    #(tranformer_name,tranformer_type,data's)
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
X_train_encoded=transformer.fit_transform(X_train)
X_test_encoded=transformer.fit_transform(X_test)
```
