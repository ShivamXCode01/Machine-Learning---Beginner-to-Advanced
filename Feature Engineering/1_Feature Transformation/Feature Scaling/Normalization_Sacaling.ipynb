**2.Normalization:-** Normalization is a feature scaling technique where the values of a dataset are rescaled to a fixed range for all the columns(features) data.

**There are major 4 types of normalization** 1.Min-Max Scaling[0,1]  ----------------  2.Mean Normalization [-1,1] -----------------3.Max Absolute scaling[-1,1]  ------------------ 4.Robust Sacling

**1.Min-Max Scaling :- Xi = (Xi - Xmin)/(Xmax-Xmin)**

```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('salary_prediction_data.csv')
df
```

```python
df.info()
```

```python
df.drop(['Education','Location','Job_Title','Gender'],axis=1,inplace=True)
```

```python
df
```

```python
# now we split or data into train and test 
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test = train_test_split(df,test_size=0.2,random_state=42)
```

```python
X_train
```

```python
X_test
```

**AAM ZINDAGI**

```python
Age_min = np.min(df['Age'])
Age_max = np.max(df['Age'])
Salary_min = np.min(df['Salary'])
Salary_max = np.max(df['Salary'])
experience_min = np.min(df['Experience'])
experience_max = np.max(df['Experience'])
```

```python
Age_min
```

```python
df_scaled = pd.DataFrame()
```

```python
df_scaled['Age_Scaling'] = (df['Age'] -Age_min)/(Age_max-Age_min)
```

```python
df_scaled['Salary_Sacaling'] = (df['Salary'] -Salary_min)/(Salary_max-Salary_min)
```

```python
df_scaled['Experience_Scaling'] = (df['Experience'] - experience_min)/(experience_max-experience_min)
```

```python
df_scaled
```

```python
print("Before Scaling")
df.drop(['Experience_Scaling','Salary_Sacaling','Age_Scaling'],inplace=True,axis = 1)
df
```

```python
print("After Scaling")
df_scaled
```

**2.Mean Normalization**
**AAM ZINDAGI**

```python
df
```

```python
Age_mean = np.mean(df['Age'])
salary_mean = np.mean(df['Salary'])
Experience_mean = np.mean(df['Experience'])
```

```python
df_mean_scaled = pd.DataFrame()
df_mean_scaled['Age_mean_Scaled'] = (df['Age'] - Age_mean )/(Age_max -Age_mean)
df_mean_scaled['Salary_mean_Scaled'] = (df['Salary'] -salary_mean) /(Salary_max-salary_mean)
df_mean_scaled['Experience_mean_Scaled'] = (df['Experience']-Experience_mean)/(experience_max-Experience_mean)
```

```python
df_mean_scaled
```

There is no class  present in the sklearn for Mean Normalization like MinMax Scalar
