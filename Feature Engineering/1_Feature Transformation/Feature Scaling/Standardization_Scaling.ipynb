**Feature Engineering:-** It is a process of extracting features from the row data using the domain knowledge.These features is used to improve the performance the machine learning models--------------------------------------------------------------------------------------------------------------------------------There are 4 major processes in the feature Engineering 1.Feature Transformation       2.Feature Construction              3.Feature selection     4.Feature Extraction

**Feature Transformation:-** 
*In Machine Learning, feature transformation is the process of converting or modifying the original features (variables) of a dataset into a new form to improve the performance and accuracy of a model.*------------------------------------------------------------------------------------------------------------------------------------------
There are many processes consists in the feature Transformation 

**1.Feature Scalling:-** it  is the process of bringing different feature values into a similar range or scale so that machine learning algorithms can work efficiently in Machine Learning.


Types of Feature Scaling ----- 1. Standarization(z-score normalization)---------------------------------         2.Normalization

**1.Standarization(z-score normalization)**:-the mean becomes 0 and the standard deviation becomes 1.
--------------Used In

Logistic Regression

Support Vector Machine

Neural Networks

```python
Age_std = np.std(df['Age'])
Age_std
```

```python
df['Age_standardized'] = (df['Age'] - Age_mean) / Age_std
```

```python
salary_mean = np.mean(df['Salary'])
```

```python
salary_std = np.std(df['Salary'])
```

```python
df['Salary_Standradized'] = (df['Salary']-salary_mean)/salary_std
```
