**DATA FRAMES**
*Data frames is nothing but it is the collection of data series*
*we are creating dataframes*

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
df
```

*We can access our desired column or series from data frames*

```python
print(df['Name'])
print(df['Salary'])
```

**Creating new column in DataFrame**

*One thing must remember during this no.of elements must be same as no. of rows*

```python
df["Profession"]=['Engi.','teacher','scietist','peon']
print(df)
```

**Removing columns from DataFrame**

```python
#df.drop('Profession',axis=1) it can't save in the memory we need implace keyword to do it permanent
df.drop('Profession',axis=1,inplace=True)
print(df)
```

**Selection of rows and columns with different method**

```python
print(df.loc[2]) #it gives row in the column format
```

```python
print(df.loc[0:2]) # here we can pass starting and ending index of the row to extracting
```

```python
print(df.loc[2:3]), [["City","Salary"]]
```

*We can also do conditional Selection*

```python
print([df["Age"] > 30])
print([df["Age"] > 30] and [df["City"]=='Paris'])
```

**Finding Missing Values**

```python
data1 = {
    'Name': ['John', None, 'Peter', 'Linda'],
    'Age': [28, 34, np.nan, 42],
    'City': [None, 'Paris', None, 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df3 =pd.DataFrame(data1)
print(df3)
```

```python
print(df3.isna()) # used to find the null values if there is any present 
                 #there null values then it gives  true else false
```

```python
print(df3.isna().sum()) #it gives the total number of null values in each row
```

```python
print(df3.isna().any())  # it gives is any null values present in row wise if any then it gives 
                        #True else False
```

**Removing Missing Values**

```python
print(df3.dropna()) # it is used to drop the row where we have null values present 
```

```python
print(df3.dropna(thresh=2))  # it is also drop the row where null present but in this we assign desired conditions that how much null values
                            # valid here we given 2 means if a row has two null values then it remove that row
                            #but in our data only one null values is present so it do not remove any row
```

**Filling Missing Value**

```python
print(df3.fillna(90)) # fillna is used to fill the null value to our desired values we want to fill
                      # here we give 90 to fill in all nulll values
```

```python
print(df3.fillna(value="Ram"))  # fillna is used to fill the null value to our desired values we want to fill
                                # here we give ram to fill in all nulll values
```

**Group By aggregation**

```python
print(df3['Age'].mean()) # we can perform multiple aggregation on a series data(columns)like mean mode median etc
print(df3['Age'].mode())
print(df3['Age'].median())
```

**Pivot Tables**

```python
print(df3)
print(pd.pivot_table(df3,values="Age",index="City",aggfunc="mean"))
print(pd.pivot_table(
    df3,
    values="Age",
    index="City",
    aggfunc=["mean", "median", "count"]
))
```
